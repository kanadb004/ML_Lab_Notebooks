### 1.

### Generate data set

```python
import pandas as pd
import numpy as np
import random
```

```python
np.random.seed(1265)
num_records = 100
```

```python
# Generate categorical data
names = [f"Student_{i}" for i in range(1, num_records + 1)]
branches = np.random.choice(['CSE', 'ECE', 'EEE', 'MECH', 'CIVIL'], num_records)

# Generate numerical data - semester, marks
semesters = np.random.choice([1, 2, 3, 4, 5, 6, 7, 8], num_records)
mark1 = np.random.randint(30, 100, num_records)
mark2 = np.random.randint(30, 100, num_records)
mark3 = np.random.randint(30, 100, num_records)
```

```python
df = pd.DataFrame({
    'Name': names,
    'Branch': branches,
    'Semester': semesters,
    'Mark_1': mark1,
    'Mark_2': mark2,
    'Mark_3': mark3
})

# Determine pass/fail - marks >= 40
df['Status'] = np.where(
    (df['Mark_1'] >= 40) & (df['Mark_2'] >= 40) & (df['Mark_3'] >= 40), 
    'Pass', 
    'Fail'
)
```

```python
df.describe()
```

#### Descriptive statistics

##### Mark 1 statistics

```python
mark1_mean = df['Mark_1'].mean()
mark1_median = df['Mark_1'].median()
mark1_mode = df['Mark_1'].mode()
mark1_var = df['Mark_1'].var()
mark1_std = df['Mark_1'].std()

print(f"Mean: {mark1_mean:.2f}, Median: {mark1_median}")

print("Mode:", end=" ")
for m in mark1_mode:
    print(m, end=" ")
print()

print(f"Variance: {mark1_var:.2f}, Standard Deviation: {mark1_std:.2f}")
```

##### Mark 2 skewness

```python
mark2_skew = df['Mark_2'].skew()

print(f"Skewness Value: {mark2_skew:.2f}")

if abs(mark2_skew) < 0.5:
    print("Distribution: Symmetric")
elif mark2_skew > 0:
    print("Distribution: Right-skewed (Positive)")
else:
    print("Distribution: Left-skewed (Negative)")
```

##### Correlation b/w Mark 2 & 3

```python
correlation = df['Mark_2'].corr(df['Mark_3'])

print(f"Correlation between Mark 2 and Mark 3: {correlation:.2f}")
```

#### Missing value handling

```python
print("Missing Values Count (Before Imputation)")
print(df.isnull().sum())
print()

# Fill missing numeric values with the column mean
numeric_cols = df.select_dtypes(include=[np.number]).columns
for col in numeric_cols:
    col_mean = df[col].mean()
    df[col].fillna(col_mean, inplace=True)

# Fill missing categorical values with the column mode
categorical_cols = df.select_dtypes(include=['object', 'category']).columns
for col in categorical_cols:
    col_mode = df[col].mode()[0]
    df[col].fillna(col_mode, inplace=True)

print("Missing Values Count (After Imputation)")
print(df.isnull().sum())
```

### Generate dataset

```python
daa = pd.DataFrame({
    'Numeric': np.random.randint(10, 100, 20),
    'Categorical': np.random.choice(['A', 'B', 'C'], 20),
    'Mixed': ['Text' if i % 3 == 0 else i for i in range(20)]
})

daa.head()
```

### Copy dataset

```python
da1 = daa.copy()

# Determine index thresholds for 25%, 50%, and 75%
n = len(da1)
q1 = n // 4
q2 = n // 2
q3 = 3 * n // 4
```

#### Apply rules

```python
for col in da1.columns:
    
    # Check if the column is purely numeric
    if pd.api.types.is_numeric_dtype(daa[col]):
        col_mean = daa[col].mean()
        col_max = daa[col].max()
        col_min = daa[col].min()
        
        col_idx = da1.columns.get_loc(col)
        da1.iloc[0:q1, col_idx] = col_mean
        da1.iloc[q1:q2, col_idx] = col_max
        da1.iloc[q2:q3, col_idx] = col_min
        da1.iloc[q3:, col_idx] = col_mean / 2

    else:        
        # Check if the column has mixed entries by trying to convert it to numeric
        numeric_converted = pd.to_numeric(daa[col], errors='coerce') # 'coerce' to turn non-numeric strings into NaN (null) values
        
        has_numbers = numeric_converted.notna().any()
        has_text = numeric_converted.isna().any()
        
        if has_numbers and has_text:
            # Mixed types - replace all non-numerical entries with zeros
            da1[col] = numeric_converted.fillna(0)
            
        else:
            # Purely non-numerical - categorical
            val_counts = daa[col].value_counts() # value_counts() to get a series with the most common values at the top
            
            most_freq = val_counts.index[0]
            least_freq = val_counts.index[-1]
            
            col_idx = da1.columns.get_loc(col)
            da1.iloc[0:q2, col_idx] = most_freq
            da1.iloc[q2:, col_idx] = least_freq
```
