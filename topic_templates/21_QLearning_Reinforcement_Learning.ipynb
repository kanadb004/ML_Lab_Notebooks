### Tabular reinforcement learning: environment, training loop, and policy evaluation

#### Define environment

```python
import numpy as np
import matplotlib.pyplot as plt

class Environment:
    """Defines state/action spaces, step() and reset() methods, and the reward function."""
    def __init__(self):
        pass

    def reset(self):
        raise NotImplementedError

    def step(self, action):
        # return next_state, reward, done
        raise NotImplementedError
```

#### Initialize Q-table & hyperparameters

```python
n_states = 10   # TODO: set to actual state-space size
n_actions = 4   # TODO: set to actual action-space size

q_table = np.zeros((n_states, n_actions))

alpha = 0.1      # learning rate
gamma = 0.9      # discount factor
epsilon = 1.0    # exploration rate
epsilon_decay = 0.995
epsilon_min = 0.01
n_episodes = 500
```

#### Training loop

```python
env = Environment()
episode_rewards = []

for episode in range(n_episodes):
    state = env.reset()
    done = False
    total_reward = 0

    while not done:
        if np.random.rand() < epsilon:
            action = np.random.randint(n_actions)
        else:
            action = int(np.argmax(q_table[state]))

        next_state, reward, done = env.step(action)

        q_table[state, action] += alpha * (
            reward + gamma * np.max(q_table[next_state]) - q_table[state, action]
        )

        state = next_state
        total_reward += reward

    epsilon = max(epsilon_min, epsilon * epsilon_decay)
    episode_rewards.append(total_reward)
```

#### Visualize training progress

```python
plt.figure(figsize=(10, 5))
plt.plot(episode_rewards)
plt.xlabel('Episode')
plt.ylabel('Total reward')
plt.title('Q-Learning Training Progress')
plt.show()
```

#### Evaluate learned policy

```python
test_env = Environment()
state = test_env.reset()
done = False
total_reward = 0

while not done:
    action = int(np.argmax(q_table[state]))
    state, reward, done = test_env.step(action)
    total_reward += reward

print('Evaluation total reward:', total_reward)
```
