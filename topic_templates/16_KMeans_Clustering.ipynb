### Partition-based clustering for numerical features

#### Load dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

df = pd.read_csv('PATH_TO_DATASET.csv')
df.head()
```

#### Select features

```python
X = df.drop(columns=['TARGET_COLUMN'])  # drop target - clustering is unsupervised
```

#### Encode / scale features

```python
scaler = StandardScaler()
X_encoded = scaler.fit_transform(X)
```

#### Find suitable number of clusters

```python
k_values = range(2, 11)
silhouette_scores = []
for k in k_values:
    km = KMeans(n_clusters=k, random_state=1265, n_init=10)
    labels = km.fit_predict(X_encoded)
    silhouette_scores.append(silhouette_score(X_encoded, labels))

plt.plot(list(k_values), silhouette_scores, marker='o')
plt.xlabel('k')
plt.ylabel('Silhouette score')
plt.title('Choosing k')
plt.show()
```

#### Train final model

```python
best_k = list(k_values)[int(np.argmax(silhouette_scores))]
kmeans_model = KMeans(n_clusters=best_k, random_state=1265, n_init=10)
cluster_labels = kmeans_model.fit_predict(X_encoded)
```

#### Performance evaluation

```python
from sklearn.metrics import silhouette_score

score = silhouette_score(X_encoded, cluster_labels)
print('Score:', score)
```

#### Visualize results

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2, random_state=1265)
X_2d = pca.fit_transform(X_encoded)
plt.scatter(X_2d[:, 0], X_2d[:, 1], c=cluster_labels, cmap='tab10', alpha=0.7)
plt.title('Cluster Visualization (PCA)')
plt.show()
```
