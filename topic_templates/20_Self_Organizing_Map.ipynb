### Unsupervised neural clustering with a self-organizing map

#### Load dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from minisom import MiniSom

df = pd.read_csv('PATH_TO_DATASET.csv')
df.head()
```

#### Select features

```python
features = ['FEATURE_1', 'FEATURE_2', 'FEATURE_3']
X = df[features].values
```

#### Feature scaling

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

#### Train SOM

```python
som_grid = 10
som = MiniSom(som_grid, som_grid, X_scaled.shape[1], sigma=1.0, learning_rate=0.5, random_seed=1265)
som.random_weights_init(X_scaled)
som.train_random(X_scaled, 1000)
```

#### Performance evaluation

```python
# Assign each sample to its winning (best matching) neuron
winners = np.array([som.winner(x) for x in X_scaled])
cluster_labels = winners[:, 0] * som_grid + winners[:, 1]
quantization_error = som.quantization_error(X_scaled)
print('Quantization error:', quantization_error)
```

#### Visualize results

```python
hit_counts = np.zeros((som_grid, som_grid))
for w in winners:
    hit_counts[w[0], w[1]] += 1

plt.figure(figsize=(7, 6))
plt.imshow(hit_counts, cmap='viridis')
plt.colorbar(label='Hit count')
plt.title('SOM Hit Map')
plt.show()
```
