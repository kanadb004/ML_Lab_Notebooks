# Descriptive Statistics & Synthetic Data Generation
## Template Notebook
### Mean, median, mode, variance, std dev, skewness, correlation, missing-value handling

#### Generate dataset

```python
import pandas as pd
import numpy as np

np.random.seed(1265)

df = pd.DataFrame({
    'Name': ['Student_' + str(i) for i in range(1, 51)],
    'Branch': np.random.choice(['CSE', 'ECE', 'MECH'], size=50),
    'Semester': np.random.randint(1, 9, size=50),
    'Mark_1': np.random.normal(70, 10, size=50),
    'Mark_2': np.random.normal(65, 15, size=50),
    'Mark_3': np.random.normal(75, 8, size=50),
})
df.head()
```

#### Descriptive statistics

```python
mean_val = df['Mark_1'].mean()
median_val = df['Mark_1'].median()
mode_val = df['Mark_1'].mode()[0]
variance_val = df['Mark_1'].var()
std_val = df['Mark_1'].std()
skew_val = df['Mark_2'].skew()
corr_val = df['Mark_2'].corr(df['Mark_3'])

print('Mean:', mean_val, 'Median:', median_val, 'Mode:', mode_val)
print('Variance:', variance_val, 'Std Dev:', std_val)
print('Skewness:', skew_val)
print('Correlation:', corr_val)
```

#### Missing value handling

```python
# Introduce missing values
mask = np.random.rand(len(df)) < 0.1
df.loc[mask, 'Mark_1'] = np.nan

print('Missing values before imputation:')
print(df.isnull().sum())

df['Mark_1'] = df['Mark_1'].fillna(df['Mark_1'].mean())

print('Missing values after imputation:')
print(df.isnull().sum())
```
