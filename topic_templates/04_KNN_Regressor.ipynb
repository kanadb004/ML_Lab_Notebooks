### KNN regression with k-selection

#### Load dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsRegressor

df = pd.read_csv('PATH_TO_DATASET.csv')
df.head()
```

#### Features & target

```python
X = df.drop(columns='TARGET_COLUMN')
y = df['TARGET_COLUMN']
```

#### Train-test split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=1265
)
```

#### Feature scaling

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

#### Find suitable k

```python
from sklearn.metrics import mean_absolute_error

k_values = range(1, 21)
mae_scores = []
for k in k_values:
    knn = KNeighborsRegressor(n_neighbors=k)
    knn.fit(X_train_scaled, y_train)
    mae_scores.append(mean_absolute_error(y_test, knn.predict(X_test_scaled)))

best_k = list(k_values)[int(np.argmin(mae_scores))]
plt.plot(list(k_values), mae_scores, marker='o')
plt.xlabel('k')
plt.ylabel('MAE')
plt.title('k vs MAE')
plt.show()
print('Best k:', best_k)
```

#### Train final model

```python
model = KNeighborsRegressor(n_neighbors=best_k)
model.fit(X_train_scaled, y_train)
```

#### Performance evaluation

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

y_pred = model.predict(X_test_scaled)
print('MAE:', mean_absolute_error(y_test, y_pred))
print('MSE:', mean_squared_error(y_test, y_pred))
print('R2:', r2_score(y_test, y_pred))
```

#### Visualize results

```python
plt.scatter(y_test, y_pred, color='teal', alpha=0.6)
plt.xlabel('Actual')
plt.ylabel('Predicted')
plt.show()

residuals = y_test - y_pred
plt.scatter(y_pred, residuals, alpha=0.6)
plt.axhline(0, color='red', linestyle='--')
plt.title('Residuals')
plt.show()
```
